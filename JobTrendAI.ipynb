```python
daily_jobs = df.groupby("posted_date").size().reset_index(name="job_count")
daily_jobs = daily_jobs.sort_values("posted_date")

daily_jobs.head()
```

```python
daily_jobs["day"] = range(len(daily_jobs))
daily_jobs.head()
```

```python
train = daily_jobs.iloc[:-30]
test = daily_jobs.iloc[-30:]

print("Training days:", len(train))
print("Testing days:", len(test))
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

x_train = train[["day"]]
y_train = train["job_count"]

model.fit(x_train, y_train)

print("Model trained successfully!")

```

```python
test["predicted_jobs"] = model.predict(test[["day"]]).round().astype(int)

test[["posted_date", "job_count", "predicted_jobs"]].head(10)
```

```python
test = test.copy()

test["predicted_jobs"] = (
    model.predict(test[["day"]])
    .round()
    .astype(int)
)

test[["posted_date", "job_count", "predicted_jobs"]].head(10)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(test["job_count"], test["predicted_jobs"])
rmse = np.sqrt(mean_squared_error(test["job_count"], test["predicted_jobs"]))

print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
```

```python
daily_jobs["rolling_7"] = daily_jobs["job_count"].rolling(7).mean()

daily_jobs[["posted_date", "job_count", "rolling_7"]].tail(10)
```

```python
model2 = LinearRegression()

train2 = daily_jobs.iloc[7:120]
test2 = daily_jobs.iloc[120:]

model2.fit(train2[["rolling_7"]], train2["job_count"])

test2 = test2.copy()
test2["predicted_jobs"] = model2.predict(test2[["rolling_7"]]).round().astype(int)

test2[["posted_date", "job_count", "predicted_jobs"]].head(10)
```

```python
mae2 = mean_absolute_error(test2["job_count"], test2["predicted_jobs"])
rmse2 = np.sqrt(mean_squared_error(test2["job_count"], test2["predicted_jobs"]))

print("Model 2 MAE:", round(mae2, 2))
print("Model 2 RMSE:", round(rmse2, 2))
```

```python
from sklearn.ensemble import RandomForestRegressor

daily_jobs["lag_1"] = daily_jobs["job_count"].shift(1)
daily_jobs["lag_7"] = daily_jobs["job_count"].shift(7)

daily_jobs = daily_jobs.dropna()

train3 = daily_jobs.iloc[:-30]
test3 = daily_jobs.iloc[-30:]

model3 = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model3.fit(
    train3[["lag_1", "lag_7"]],
    train3["job_count"]
)

test3 = test3.copy()
test3["predicted_jobs"] = model3.predict(
    test3[["lag_1", "lag_7"]]
).round().astype(int)

test3[["posted_date", "job_count", "predicted_jobs"]].head(10)
```

```python
mae3 = mean_absolute_error(test3["job_count"], test3["predicted_jobs"])
rmse3 = np.sqrt(mean_squared_error(test3["job_count"], test3["predicted_jobs"]))

print("Model 3 MAE:", round(mae3, 2))
print("Model 3 RMSE:", round(rmse3, 2))
```
